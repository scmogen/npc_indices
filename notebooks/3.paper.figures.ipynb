```python
def detrend_linear(dat, dim, order):
    """ linear detrend dat along the axis dim """
    params = dat.polyfit(dim=dim, deg=order)
    fit = xr.polyval(dat[dim], params.polyfit_coefficients)
    dat = dat-fit
    return dat, fit
```

## Load data and indices

```python
# open the data!
ds = xr.open_dataset('SSH.regrid.nc')['SSH']
ds['time'] = pd.date_range("1958-01", "2020-12", freq="MS")
# center this bad boy on the Pacific
ds = ds.roll(lon=180,roll_coords=True)
ds['lon'] = np.arange(0.5,360.5,1)
# select the NE Pacific
ds = ds/100
```

```python
bifu_cesm2 = xr.open_mfdataset('./CESM2.LoB/CESM2.Lat.Bifu.member*.nc',combine='nested',concat_dim='member').load().mean('lon_calc').bifurcation
bifu_fosi = xr.open_dataset('FOSI.Lat.Bifu.contour.nc').bifurcation.mean('lon_calc')#['bifurcation_latitude'].isel(lon=0)
bifu_mer = xr.open_dataset('SODA.Lat.Bifu.contour.nc').bifurcation.mean('lon_calc')#['line']

core_fosi = xr.open_dataset('FOSI.Lat.Core.nc')['core']
core_cesm2 = xr.open_dataset('CESM2.Lat.Core.nc')['core']
core_mer = xr.open_dataset('SODA.Lat.Core.nc')['core']

str_fosi = xr.open_dataset('FOSI.UVEL.strength.nc')['strength']
str_cesm2 = xr.open_dataset('CESM2.UVEL.strength.nc')['strength']
str_mer = xr.open_dataset('SODA.Core.strength.nc')['ugeo']

npgo_fosi = xr.open_dataset('FOSI.NPGO.nc')['PC2']
# npgo_soda = xr.open_dataset('SODA.NPGO.nc')['PC2']
npgo_obs =  xr.open_dataset('obs.NPGO.nc')['NPGO index']

pdo_fosi = -xr.open_dataset('FOSI.PDO.nc')['PC1']
# pdo_soda = xr.open_dataset('SODA.PDO.nc')['PC1']
pdo_obs = xr.open_dataset('pdo.timeseries.sstens.nc')['pdo']
```

```python
from scipy.stats import linregress

def process(ds, trend, order):
    ds_proc = ds.groupby('time.month') - ds.groupby('time.month').mean()
    ds_proc = detrend_linear(ds_proc,'time',order)
    return ds_proc
```

```python
# CORRELATE WITH INDICES OF FOSI

# might need to add ufunc!

def apply_lin_regress(index, variable):
    range = np.arange(index.min(), index.max(), (index.max() - index.min()) / index.size)
    m, b, r, p, err = linregress(index, variable)
    return r, p

def calc(da, x, coord='time'):
    """Finds the correlation along a given dimension of a dataarray."""

    return xr.apply_ufunc(apply_lin_regress, 
                          da, 
                          x,
                          input_core_dims=[[coord],[coord]] , 
                          output_core_dims=[[],[]],
                          vectorize=True,
                          output_dtypes=[float, float]
                          )

# when you want to run code, use: r,p = calc(core_fosi, surf_vars.O2) for a given tracers
```
